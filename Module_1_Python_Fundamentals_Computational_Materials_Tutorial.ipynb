1. Remove by Value:

remove(): The remove() method deletes the first occurrence of a specific value from the list.

Warning: If the value does not exist, Python raises a ValueError.

2. Remove by Index:

pop() : The pop() method removes an item at a specific index and returns it. If you do not provide an index, it removes and returns the last item.

3. Remove by Index or Slice:

del: The del keyword deletes an item or a slice range at a specific position but does not return any value.

4. Remove All Items:

clear() : The clear() method removes every element from the list, leaving it completely empty.

````python
fruits = ["apple", "banana", "cherry", "banana"]
fruits.remove("banana")
print(fruits)
# Output: ['apple', 'cherry', 'banana']
````

````python
numbers = [10, 20, 30, 40]

# Remove by index
removed_item = numbers.pop(1)
print(removed_item)  # Output: 20
print(numbers)       # Output: [10, 30, 40]

# Remove the last item
last_item = numbers.pop()
print(last_item)     # Output: 40
````

````python
letters = ["a", "b", "c", "d", "e"]

# Delete a single item
del letters[0]
print(letters)  # Output: ['b', 'c', 'd', 'e']

# Delete a slice range (index 1 up to but excluding 3)
del letters[1:3]
print(letters)  # Output: ['b', 'e']
````

````python
items = [1, 2, 3]
items.clear()
print(items)  # Output: []
````

### Scientific example — Arrhenius equation

`D = D0 exp(-Q / RT)`

Calculate a diffusion coefficient at several temperatures.


````python
import math

D0 = 1e-5       # m^2/s
Q = 80000       # J/mol
R = 8.314       # J/(mol K)

temperatures = [600, 700, 800, 900, 1000]

for T in temperatures:
    D = D0 * math.exp(-Q / (R * T))
    print(f"T = {T:4d} K, D = {D:.3e} m^2/s")
````

# 19. Functions with multiple outputs

Python functions can return multiple values.

Example: calculating stress and strain energy density.


````python
def stress_energy(force, area, displacement, original_length):
    stress = force / area
    strain = displacement / original_length
    energy_density = 0.5 * stress * strain
    return stress, strain, energy_density

stress, strain, energy = stress_energy(
    force=10000,
    area=1e-4,
    displacement=0.001,
    original_length=0.1
)

print("Stress =", stress, "Pa")
print("Strain =", strain)
print("Energy density =", energy, "J/m^3")
````

# 27. Basic programming pattern: parameter sweep

A common computational-materials task is:

**choose parameters → run model → store result → analyze results**

We can implement the structure even before using NumPy.


````python
import math

def diffusion_coefficient(T, D0, Q):
    if T <= 0:
        raise ValueError("Temperature must be positive.")
    if Q < 0:
        raise ValueError("Activation energy must be non-negative.")

    R = 8.314
    return D0 * math.exp(-Q / (R * T))

temperatures = range(500, 1201, 100)

results = []

for T in temperatures:
    D = diffusion_coefficient(T, 1e-5, 80000)
    results.append((T, D))

for T, D in results:
    print(f"{T} K -> {D:.3e} m^2/s")
````

# 34. Challenge — Implement a root-finding algorithm

Before using SciPy, implement the **bisection method** using only Python fundamentals.

Given a function `f(x)` and an interval `[a,b]` where the function changes sign:

1. calculate the midpoint,
2. evaluate the function,
3. select the appropriate half interval,
4. repeat until the interval is sufficiently small.

This is an important bridge from Python fundamentals to numerical methods.


The Bisection Method is a simple numerical algorithm used to find a zero (root) of a continuous function, \(f(x) = 0\). It is based on the Intermediate Value Theorem and works by repeatedly cutting a search interval in half.

Core Concept: How It Works

Find an Interval: You start with two points, \(a\) and \(b\), where the function changes signs.

This means \(f(a)\) and \(f(b)\) must have opposite signs (\(f(a) * f(b) < 0\)).

This guarantees a root exists between them.
Find the Midpoint: Calculate the middle point of the interval:

\(c= ({a+b}/{2}))

Check the Midpoint:

Check the left half: If f(a) * f(c) < 0, the root is between a and c. Set b = c.

Check the right half: If f(b) * f(c) < 0, the root is between c and b. Set a = c.

Continue till a tolerance of 10^6 or maximum of 100 iterations whichever is earlier. Take any polynomial or function of your choice.

````python
def bisection(f, a, b, tolerance=1e-6, max_iterations=100):
    fa = f(a)
    fb = f(b)

    if fa * fb > 0:
        raise ValueError(
            "Function must have opposite signs at a and b."
        )

    for iteration in range(max_iterations):
        c = (a + b) / 2
        fc = f(c)

        if abs(fc) < tolerance or abs(b - a) < tolerance:
            return c, iteration + 1

        if fa * fc < 0:
            b = c
            fb = fc
        else:
            a = c
            fa = fc

    return (a + b) / 2, max_iterations


def f(x):
    return x**2 - 2

root, iterations = bisection(f, 0, 2)

print("Root:", root)
print("Iterations:", iterations)
print("Check:", f(root))
````
